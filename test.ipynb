```python
def pretokenize(text: str) :
    text_removed_special_tokens = remove_special_tokens(text)
    PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""

    pre_tokens = {}
    for match in re.finditer(PAT, text_removed_special_tokens):
        tok = match.group(0)
        pre_tokens[tok] = pre_tokens.get(tok, 0) + 1

    return pre_tokens
```

```python

import multiprocessing as mp
import regex as re
from typing import List, Dict, Tuple

from cs336_basics.pretokenization_example import find_chunk_boundaries

NUM_PROCESSES = 4
chunk_token = "<|endoftext|>"
special_tokens = ["<|endoftext|>"]
file_path = r"E:\LLM\CS336\assignment1-basics\data\TinyStoriesV2-GPT4-valid.txt"

def pretokenize_file(file_path: str, chunk_special_token: str, special_tokens):
    def process_chunk(start_end):
        start, end = start_end

        # Each process opens the file separately
        with open(file_path, "rb") as f:
            f.seek(start)
            chunk = f.read(end - start).decode("utf-8", errors="ignore")

        pre_token_counts = pretokenize(chunk, special_tokens)

        return pre_token_counts
    
    chunk_token_bytes: bytes = chunk_special_token.encode("utf-8")
    with open(file_path, "rb") as f:
        boundaries = find_chunk_boundaries(f, NUM_PROCESSES, chunk_token_bytes)
        chunk_starts_ends = list(zip(boundaries[:-1], boundaries[1:]))

    # Parallel execution
    with mp.Pool(NUM_PROCESSES) as pool:
        results = pool.map(process_chunk, chunk_starts_ends)

    return results

print(pretokenize_file(file_path, chunk_token, special_tokens))
```
